# Notes on Neutrino Oscillations in a Vacuum

### Assumptions:

#### Neutrinos are not Majorana, they do not violate CP symmetry, only considering neutrinos and not anti-neutrinos

Neutrino Eigenstates relative to one another are quantified by the value $\ket{\nu_i} = \sum_{\alpha}U^{*}_{\alpha i}\ket{\nu_{\alpha}}$, and $\ket{\nu_\alpha} = \sum_{\alpha}U_{\alpha i}\ket{\nu_{i}}$ where $\ket{\nu_\alpha}$ is a neutrino with definite flavour $\alpha = e, \mu, \tau$, and $\ket{\nu_i}$ is a neutrino with definite mass $m_i$, $i = 1,2,3$. The U represents the complex conjugate; for antineutrinos, the complex conjugate should be removed from the first equation and inserted into the second.

$U_{\alpha i}$ represents the PMNS matrix, and represents the matrix elements of the matrix
$$
U = \begin{bmatrix}
U_{e1} & U_{e2} & U_{e3} \\
U_{\mu 1} & U_{\mu 2} & U_{\mu 3} \\
U_{\tau 1} & U_{\tau 2} & U_{\tau 3} 
\end{bmatrix}
$$

$$
U=
    \begin{bmatrix}
        c_{12}c_{13} & s_{12}c_{13} & c_{13}e^{-i\delta} \\
        -s_{12}c_{23} - c_{12}s_{23}s_{13}e^{i\delta} & -c_{12}c_{23} - s_{12}s_{23}s_{13}e^{i\delta} & s_{23}c_{13} \\
        s_{12}s_{23} - c_{12}c_{23}s_{13}e^{i\delta} & -c_{12}s_{23} - s_{12}c_{23}s_{13}e^{i\delta} & c_{23}c_{13} 
    \end{bmatrix}
    \begin{bmatrix}
        e^{\frac{i\alpha_1}{2}} & 0 & 0 \\
        0 & e^{\frac{i\alpha_2}{2}} & 0 \\
        0 & 0 & 1 
    \end{bmatrix}
$$

Where $c_{ij} = cos(\theta_{ij})$, and $s_{ij} = sin(\theta_{ij})$, and $\theta_{ij} = \theta_{ji}$. Phase factors $\alpha_{1},\alpha_{2}$ are only necessary if we consider these neutrinos as majorana neutrinos, which we can make the simplifying assumption that they are not, collapsing the U matrix into

$$
U=
    \begin{bmatrix}
        cos(\theta_{12})cos(\theta_{13}) & sin(\theta_{13})cos(\theta_{13}) & cos(\theta_{13})e^{-i\delta} \\
        -sin(\theta_{13})sin(\theta_{23}) - cos(\theta_{12})sin(\theta_{23})sin(\theta_{13})e^{i\delta} & -cos(\theta_{12})sin(\theta_{23}) - sin(\theta_{13})sin(\theta_{23})sin(\theta_{13})e^{i\delta} & sin(\theta_{23})cos(\theta_{13}) \\
        sin(\theta_{13})sin(\theta_{23}) - cos(\theta_{12})sin(\theta_{23})sin(\theta_{13})e^{i\delta} & -cos(\theta_{12})sin(\theta_{23}) - sin(\theta_{13})sin(\theta_{23})sin(\theta_{13})e^{i\delta} & sin(\theta_{23})cos(\theta_{13}) 
    \end{bmatrix}
$$

Through experimental measurements, we know that $\theta_{12} = 33.41 ^{\circ +0.75^\circ}_{\text{ }-0.72^\circ}$, $\theta_{13} = 8.54^{\circ +0.11^\circ}_{\text{ }-0.12^\circ}$, $\theta_{23} = 49.1 ^{\circ +1.0^\circ}_{\text{ }-1.3^\circ}$, $\delta_{CP} = 197 ^{\circ +42^\circ}_{\text{ }-0.25^\circ}$

As of Nov. 2022, the 3$\sigma$ ranges for the magnitudes of the elements of the matrix were (arXiv:2007.14792)
$$ U = \begin{bmatrix}
0.803 \sim 0.845 & 0.514 \sim 0.578 & 0.142 \sim 0.155 \\
0.233 \sim 0.505 & 0.460 \sim 0.693 & 0.630 \sim 0.779 \\
0.262 \sim 0.525 & 0.473 \sim 0.702 & 0.610 \sim 0.762 
\end{bmatrix}  $$


Eigenstates with different masses propagate with different frequencies. The heavier ones oscillate faster compared to the lighter ones. Since the mass eigenstates are combinations of flavor eigenstates, this difference in frequencies causes interference between the corresponding flavor components of each mass eigenstate. Constructive interference causes it to be possible to observe a neutrino created with a given flavor to change its flavor during its propagation. The probability that a neutrino originally of flavor $\alpha$ will later be observed as having flavor $\beta$ is
$$P_{\alpha \rightarrow \beta} = |\bra{\nu_{\beta}}\ket{\nu_{\alpha}(L)}|^2 = \left|\sum_{j}U_{\alpha,j}^* U_{\beta,j}e^{-im^2_j L/(2E)}\right|^2$$

This can be written as $$P_{\alpha \rightarrow \beta} = \delta_{\alpha \beta} - 4 \sum_{j>k}Re\{U^*_{\alpha j}U_{\beta j}U_{\alpha k}U^{*}_{\beta k}\}sin^2\left(\frac{\Delta_{jk}m^2L}{4E}\right) + 2 \sum_{j>k}Im\{U^*_{\alpha j}U_{\beta j}U_{\alpha k}U^{*}_{\beta k}\}sin\left(\frac{\Delta_{jk}m^2L}{4E}\right)$$
Where $\Delta_{jk} m^2 = m_j^2 - m_k^2$
The phase which is responsible for oscillation is often written as
$$\frac{\Delta_{jk}m^2L}{4E}\approx 1.27 \times \frac{\Delta_{jk}m^2}{eV^2}\frac{L}{km}\frac{GeV}{E}$$
The mass differences, $\Delta m^2$ are known to be of order $10^{-4}(\text{eV/c}^2)^2 = (12^{-2}\text{eV/c}^2)^2$. Oscillation distances, L, are on the order of Kilometers, and neutrino energies E are typically on order of MeV or GeV

We can make the simplifying assumption of no CP-violation to make $\delta = 0$ which will zero out the second, imaginary sum. This leaves us with the probability of transferring from neutrino $\alpha \rightarrow \beta$ as 
$$P_{\alpha \rightarrow \beta} = - 4 \sum_{j>k}Re\{U^*_{\alpha j}U_{\beta j}U_{\alpha k}U^{*}_{\beta k}\}sin^2\left(\frac{\Delta_{jk}m^2L}{4E}\right)$$

We can now arrive at the two-neutrino case which may use in our project, assuming that the neutrino will mix from one particle to another with little interference from a third, we can write the mixing matrix as
$$ U= \begin{bmatrix}
cos(\theta) & sin(\theta) \\
-sin(\theta) & cos(\theta) 
\end{bmatrix}  $$

With probability of neutrino mixing as

$$P_{\alpha\rightarrow\beta , \alpha \neq \beta} = \text{sin}^2(2\theta)\text{sin}^2\left(\frac{\Delta m^2 L}{4E}\right)$$

Or with SI units instead of natural units we get

$$P_{\alpha\rightarrow\beta , \alpha \neq \beta} = \text{sin}^2(2\theta)\text{sin}^2\left(1.27\frac{\Delta m^2 L}{E}\frac{[\text{eV}^2][\text{km}]}{[\text{GeV}]}\right)$$

# Notes on Neutrino Oscillations in a Medium

Main text will come from here https://arxiv.org/pdf/hep-ph/0412391 and https://docs.neutrino.xyz/matter/msw.html

The Hamiltonian for Matter Effect is
$$\vec{H}(x) = \left(\frac{\lambda(x)}{2} - \frac{\omega_v}{2}\text{cos}2\theta_v \right) \vec{\sigma_{3}} + \frac{\omega_v}{2}\text{sin}2\theta_v\vec{\sigma_1}$$
where $\lambda(x) = \sqrt{2}G_F n_e(x)$ is the matter potential and $\omega_v = \frac{\Delta m^2}{2E}$ is the vacuum oscillation angular frequency. $G_F$ is the Fermi coupling constant and $n_e$ is the number density of electrons in the medium. Solar Core has electron number density of $\approx 10^{37}m^{-3}$, while the outer core has $\approx 10^{20}m^{-3}$. We can create a distribution of electron neutrino oscillations from the sun to the earth which will start from the core and move to the outer edge and to earth.

We can define a new oscillation frequency $\omega_m$ and mixing angle $\theta_m$ $$\omega_m = \omega_v \sqrt{\left(\frac{\lambda}{\omega_v} - \text{cos}2\theta_{v} \right)^2 + \text{sin}^2 2\theta_v}$$ where $\lambda = \omega_v \text{cos}2\theta_v$

For relativistic neturinos propagating through space the evolution of the flavor state vector $\ket{\nu_{\alpha}(t)} = \sum U_{\alpha i}\ket{\nu_i}$ is controlled by the Hamiltonian H. In the mass basism the schrodinger-like equation is
$$i \frac{d}{dt} \binom{\nu_e}{\nu_{\mu}} = \left[ \frac{1}{2E} U \begin{bmatrix} m_{1}^2 & 0 \\ 0 & m^2_{2} \end{bmatrix} U^{\dagger} + V_{matter} \right] \binom{\nu_e}{\nu_\mu}$$
Matter potential $V_{matter}$ represents coherent forward scattering off electrons via the MSW effect

```python
sinsteps = 100  # x direction
xsteps = 100  # y direction

sintt = 10**np.linspace(-3.7, 0, sinsteps)  # Mixing angle
Rsun = 6.96342e8  # Radius of the Sun [m]
hbar = 6.58211928e-16  # Planck constant [eV*s]
c = 2.99792e8  # Speed of light [m/s]

# Calculate survival probability for different mixing angles and mixing parameters
P = np.zeros([xsteps, sinsteps])
s=0
ydata = []
for i in sintt:
    costotheta = np.cos(np.arcsin(np.sqrt(i)))
    index = -1
    for x in 10**np.linspace(-1, 6, xsteps):
        if np.shape(ydata)[0] < xsteps:
            ydata.append(x)
        costothetatilde = (-x + costotheta) / np.sqrt((x - costotheta)**2 + i)
        gamma = Rsun / (2 * x * 10.5) * i / costotheta * (1.52e-5 / 1e6) * 1 / (hbar * c)
        Pf = np.exp(-np.pi * .5 * gamma)
        Pv = .5 + (.5 - Pf) * costotheta * costothetatilde
        P[index,s] = Pv
        index -= 1
    s+=1`

# Turn the results into a heatmap
X, Y = np.meshgrid(sintt, ydata[::-1])
plt.pcolor(X, Y, P, cmap="jet")

# Styling the figure
plt.colorbar().set_label(r'Survival probability, $\nu_e\rightarrow\nu_e$')
plt.gca().set_yscale('log')
plt.gca().set_xscale('log')
plt.axis([sintt[0], sintt[-1], .1, 1e6])
plt.title('The MSW triangle')
plt.xlabel(r'$\sin^22\theta$')
plt.ylabel(r'$A/\Delta m^2$')
plt.gca().invert_yaxis()
```
